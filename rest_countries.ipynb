```python
import requests
import pandas as pd
import json

def country_data():
    
    url =  "https://restcountries.com/v3.1/all?fields=name,population,area,region"
    
    # Task 1 — Fetch and Validate the Response
    try:
        response = requests.get(url,timeout=20)
        print(f"HTTP Status Code: {response.status_code}")
        response.raise_for_status()  # Raise an exception for HTTP errors
    

        if response.status_code == 200:
            raw_data = response.json()
            print(f"Total number of countries retrieved: {len(raw_data)}")

            # Task 2 — Extract and Display Key Fields
            countries_list = []
            for country in raw_data[]:  # Display only the first 10 countries for brevity
                countries_list.append({
                    "common_name": country.get("name", {}).get("common", "N/A"),
                    "population": country.get("population", 0.0),
                    "area": country.get("area", 0.0),
                    "region": country.get("region", "N/A")
                })
                
                df = pd.DataFrame(countries_list)
                
                print(df.head(10).to_string(index=False))
        
            region_pop= df.groupby("region")["population"].sum()
            print("\n -----Task 3: Total population by region-----\n")
            print(region_pop)

            # Task 3 — Summarize the Data
            highest_pop_region = region_pop.idxmax()
            print("\n -----Task 4: Region with the highest population-----\n")
            print(highest_pop_region)

            largest_area_region = df.groupby("region")["area"].sum().idxmax()
            print("\n -----Task 5: Region with the largest area-----\n")
            print(largest_area_region)
        
        else:
            print(f"Failed to retrieve data. HTTP Status Code: {response.status_code}")
    except requests.exceptions.RequestException as e:
            print(f"An error occurred while fetching data: {e}")
    
if __name__ == "__main__":
    country_data()
        
```
